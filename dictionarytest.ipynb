```python
# Create a dictionary representing a laptop
# A dictionary stores data as key-value pairs

laptop = {
    "brand": "Dell",
    "model": "XPS 15",
    "price": 120000,
    "ram": "16GB",
    "storage": "512GB SSD",
    "available": True
}

# Display the complete laptop dictionary
print("Laptop:", laptop)
```

```python
# Access the brand using its key
print("Brand:", laptop["brand"])

# Access the model using its key
print("Model:", laptop["model"])

# Access the price using its key
print("Price:", laptop["price"])
```

```python
# Change the value of the price key
laptop["price"] = 110000
```

```python
laptop["processor"] = "Intel Core i7"
```

```python
print("After adding processor:", laptop)
```

```python
# Add GPU information to the dictionary
laptop["gpu"] = "NVIDIA RTX 4050"

print("After adding GPU:", laptop)
```

```python
# Update the existing RAM value
laptop["ram"] = "32GB"

print("Updated RAM:", laptop["ram"])
```

```python
# del removes the key and its value from the dictionary
del laptop["available"]

print("After removing available:", laptop)
```

```python
# keys() returns all the keys in the dictionary
print("All Keys:", laptop.keys())
```

```python
# values() returns all the values in the dictionary
print("All Values:", laptop.values())
```

```python
# items() returns all key-value pairs
print("All Items:", laptop.items())
```
