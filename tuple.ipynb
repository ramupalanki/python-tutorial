```python
# Create a tuple containing different programming technologies
# Notice that "Python" appears multiple times
technologies = ("Python", "Java", "Python", "C++", "JavaScript", "Python")

# Print the tuple
print("Tuple:", technologies)
```

```python
# type() tells us what type of data we have
print("Type:", type(technologies))
```

```python
# Index 0 represents the first item
print("First item:", technologies[0])
```

```python
# Index -1 represents the last item
print("Last item:", technologies[-1])
```

```python
# Slicing allows us to get a portion of the tuple
# [1:4] means start at index 1 and stop before index 4
sliced_tuple = technologies[1:4]

print("Sliced tuple:", sliced_tuple)
```

```python
# count() tells us how many times "Python" appears
python_count = technologies.count("Python")

print("Python appears:", python_count, "times")
```

```python
# index() returns the position of the first matching item
cpp_index = technologies.index("C++")

print("Index of C++:", cpp_index)
```

```python
# len() tells us how many items are in the tuple
tuple_length = len(technologies)

print("Tuple length:", tuple_length)
```

```python
# list() converts the tuple into a list
# A list is mutable, so we can change it
technology_list = list(technologies)

print("Converted to list:", technology_list)
```

```python
# append() adds "Go" to the end of the list
technology_list.append("Go")

print("List after adding Go:", technology_list)
```

```python
# tuple() converts the list back into a tuple
technologies = tuple(technology_list)

print("Final tuple:", technologies)
```
