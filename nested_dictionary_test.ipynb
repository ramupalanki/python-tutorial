```python
# Create an employee dictionary
# The "skills" key contains another dictionary
# This is called a nested dictionary

employee = {
    "name": "Amit",
    "department": "Engineering",
    "skills": {
        "language": "Python",
        "database": "PostgreSQL",
        "cloud": "AWS"
    },
    "salary": 80000
}

# Print the complete employee dictionary
print("Employee:", employee)
```

```python
# Access the "name" value using its key
print("Employee Name:", employee["name"])
```

```python
# Access the "department" value
print("Department:", employee["department"])
```

```python
# "skills" contains another dictionary
# We can access the complete nested dictionary using ["skills"]
print("Skills:", employee["skills"])
```

```python
# First access the "skills" dictionary
# Then access the "language" key inside it
print("Programming Language:", employee["skills"]["language"])
```

```python
# Access the "database" key inside the nested skills dictionary
print("Database:", employee["skills"]["database"])
```

```python
# Access the "cloud" key inside the nested skills dictionary
print("Cloud Technology:", employee["skills"]["cloud"])
```

```python
# Update the value of "language"
# "language" is inside the "skills" dictionary
employee["skills"]["language"] = "Python + JavaScript"

print("Updated Language:", employee["skills"]["language"])
```

```python
# Update the employee's salary
employee["salary"] = 90000

print("Updated Salary:", employee["salary"])
```

```python
# Add a new key-value pair to the main employee dictionary
employee["experience"] = 3

print("Experience:", employee["experience"])
```

```python
# Add a new skill inside the nested "skills" dictionary
employee["skills"]["framework"] = "Django"

print("Updated Skills:", employee["skills"])
```
