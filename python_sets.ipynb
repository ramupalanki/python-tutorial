```python
# Create a list containing some duplicate numbers
numbers = [10, 20, 30, 20, 40, 10, 50, 30, 60]
```

```python
# Print the original list
print("Original list:", numbers)
```

```python
# Convert the list into a set
# A set automatically removes duplicate values
numbers_set = set(numbers)
```

```python
# Print the set
print("Set:", numbers_set)

# Notice that duplicate values such as 10, 20, and 30
# appear only once in the set
```

```python
# Convert the set back into a list
unique_numbers = list(numbers_set)
```

```python
# Print the new list containing only unique values
print("List after removing duplicates:", unique_numbers)
```

```python
# Find the number of elements in the original list
original_count = len(numbers)
```

```python
# Print the number of original elements
print("Number of original elements:", original_count)
```

```python
# Find the number of unique elements
unique_count = len(numbers_set)
```

```python
# Print the number of unique elements
print("Number of unique elements:", unique_count)
```
