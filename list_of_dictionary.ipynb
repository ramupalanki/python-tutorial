```python
# Create a list containing three dictionaries
# Each dictionary represents one product

products = [
    {
        "name": "Laptop",
        "price": 70000,
        "brand": "Dell"
    },
    
    {
        "name": "Phone",
        "price": 40000,
        "brand": "Samsung"
    },
    
    {
        "name": "Tablet",
        "price": 30000,
        "brand": "Apple"
    }
]

# Print the complete list of products
print("Products:", products)
```

```python
# Print the complete product dataset
print("All Products:")
print(products)
```

```python
# Index 0 represents the first dictionary in the list
print("First Product:")
print(products[0])
```

```python
print("Second Product Price:")
print(products[1]["price"])
```

```python
# Index 2 represents the third product
# ["brand"] accesses the brand value

print("Third Product Brand:")
print(products[2]["brand"])
```

```python
# products[0] accesses the first product
# ["price"] accesses its price
# We then assign a new price

products[0]["price"] = 65000

print("Updated First Product:")
print(products[0])
```

```python
# Access the second product using index 1
# Add a new key called "rating"

products[1]["rating"] = 4.5

print("Updated Second Product:")
print(products[1])
```

```python
# Add a new dictionary to the end of the products list
# append() adds one new product

products.append({
    "name": "Smartwatch",
    "price": 15000,
    "brand": "Apple"
})

print("After adding another product:")
print(products)
```

```python
# Print the complete dataset after all changes

print("Final Product Dataset:")
print(products)
```
