```python
message = "   Welcome To Python Programming Class   "
```

```python
#strip() method removes extra whitespace from the beginning and end
print(message.strip())
```

```python
#lower() method converts the entire text into lowercase
print(message.lower())
```

```python
#upper() method converts the entire text into uppercase
print(message.upper())
```

```python
#title() method converts first character of each word into uppercase
print(message.title())
```

```python
#replace() method replaces given text in the string if exist
print(message.replace("Python","Advanced Python"))
```

```python
# startswith() method checks whether string starts with given text or not
print(message.strip().startswith("Welcome"))
```

```python
#endswith() method checks whether string ends with the given text or not
print(message.strip().endswith("Class"))
```

```python
#count() method gives the count of the text given in the string
print(message.count("o"))
```

```python
# find() method identifies the position of the text in the variable

print(message.find("Programming"))
```

```python
# Split() method converts sentence into words separated by whitespaces by default
print(message.split())
```
