```python
# Create a nested list containing student records
students = [
    ["Rahul", 21, "Python"],
    ["Priya", 22, "Data Science"],
    ["Aman", 20, "Machine Learning"]
]
```

```python
# 1. Print the complete list
# Purpose: Display the entire nested list
print(students)
```

```python
# 2. Print Rahul's name
# First [0] selects Rahul's record.
# Second [0] selects Rahul's name from that record.
print(students[0][0])
```

```python
# 3. Print Priya's age
# First [1] selects Priya's record.
# Second [1] selects Priya's age from that record.
print(students[1][1])
```

```python
# 4. Print Aman's course
# First [2] selects Aman's record.
# Second [2] selects Aman's course from that record.
print(students[2][2])
```

```python
# 5. Print the complete record of Priya
# [1] selects the complete second record.
print(students[1])
```

```python
# 6. Change Rahul's course to "AI"
# [0] selects Rahul's record.
# [2] selects the course inside Rahul's record.
students[0][2] = "AI"
print(students)
```

```python
# 7. Add another student record manually
# append() adds a new item to the end of the list.
students.append(["Sneha", 23, "Python"])
print(students)
```
