```python
# Create a dictionary containing student information
student = {
    "name": "Rahul",
    "age": 22,
    "course": "Python",
    "city": "Bangalore",
    "marks": 88
}

# Print the complete dictionary
print("Student:", student)
```

```python
# Access the value of "name" using its key
print("Student Name:", student["name"])
```

```python
# Access the value of "course"
print("Course:", student["course"])
```

```python
# keys() returns all the keys in the dictionary
print("Keys:", student.keys())
```

```python
# values() returns all the values
print("Values:", student.values())
```

```python
# items() returns all key-value pairs
print("Key-Value Pairs:", student.items())
```

```python
# Update the value of the "marks" key
student["marks"] = 92

print("Updated Marks:", student["marks"])
```

```python
# Add a new key called "email"
student["email"] = "rahul@example.com"

print("After adding email:", student)
```

```python
# Add a new phone number
student["phone"] = "9876543210"

print("After adding phone:", student)
```

```python
# pop() removes the specified key and returns its value
removed_city = student.pop("city")

print("Removed City:", removed_city)
print("After removing city:", student)
```

```python
# get() retrieves the value associated with a key
student_name = student.get("name")

print("Student Name:", student_name)
```

```python
# copy() creates a separate copy of the dictionary
student_copy = student.copy()

print("Original Dictionary:", student)
print("Copied Dictionary:", student_copy)
```

```python
# update() can change existing values
# and add new key-value pairs
student.update({
    "marks": 95,
    "course": "Advanced Python"
})
print("Original Dictionary:", student)
```
