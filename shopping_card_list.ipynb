```python
# Create a shopping cart using a Python list
cart = ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones"]
```

```python
# Display the original cart
print("Original Cart:", cart)
```

```python
# Access the first product
# List indexing starts from 0
print("First Product:", cart[0])

# Access the last product
# -1 represents the last item in a list
print("Last Product:", cart[-1])
```

```python
# append() adds one item to the END of the list
cart.append("Webcam")

print("After adding Webcam:", cart)
```

```python
# insert() adds an item at a specific position
# Here, "USB Hub" is inserted at index 2
cart.insert(2, "USB Hub")

print("After inserting USB Hub:", cart)
```

```python
# remove() removes the first matching item
# Here, it removes "Mouse" from the cart
cart.remove("Mouse")

print("After removing Mouse:", cart)
```

```python
# pop() removes and returns an item
# Without an index, pop() removes the LAST item
removed_item = cart.pop()

print("Removed Item:", removed_item)
print("Cart after pop():", cart)
```

```python
# index() returns the position of a specific item
monitor_index = cart.index("Monitor")

print("Index of Monitor:", monitor_index)
```

```python
# count() tells us how many times an item appears
laptop_count = cart.count("Laptop")

print("Laptop Count:", laptop_count)
```

```python
# copy() creates a separate copy of the list
cart_copy = cart.copy()

print("Original Cart:", cart)
print("Copied Cart:", cart_copy)
```

```python
# reverse() reverses the order of items
# It changes the original list
cart.reverse()

print("Reversed Cart:", cart)
```

```python
# sort() arranges the items in alphabetical order
# It changes the original list
cart.sort()

print("Sorted Cart:", cart)
```
