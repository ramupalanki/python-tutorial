```python
# Create a list containing marks of seven students
marks = [78, 85, 90, 67, 88, 92, 76]

# Print the complete list
print("Complete list:", marks)
```

```python
# Index 0 represents the first element
print("First element:", marks[0])
```

```python
# Index -1 represents the last element
print("Last element:", marks[-1])
```

```python
# Slicing [2:6] gets elements from index 2
# up to, but not including, index 6
print("Elements from index 2 to 5:", marks[2:6])
```

```python
# len() returns the number of elements in the list
print("Number of elements:", len(marks))
```

```python
# min() returns the smallest value in the list
print("Minimum marks:", min(marks))
```

```python
# sum() adds all the values in the list
print("Total marks:", sum(marks))
```

```python
# sort() arranges the list from smallest to largest
marks.sort()

print("Ascending order:", marks)
```

```python
# sort(reverse=True) arranges the list from largest to smallest
marks.sort(reverse=True)

print("Descending order:", marks)
```

```python
# append() adds one item to the end of the list
marks.append(95)

print("After adding 95:", marks)
```

```python
# extend() adds multiple items to the list
marks.extend([81, 84])

print("After adding 81 and 84:", marks)
```

```python
# remove() removes the first occurrence of the specified value
marks.remove(67)

print("After removing 67:", marks)
```

```python
# count() tells us how many times 90 appears
print("Number of times 90 occurs:", marks.count(90))
```

```python
# index() returns the position of the first occurrence of 88
print("Index of 88:", marks.index(88))
```
