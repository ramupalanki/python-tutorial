```python
# Create a set of students learning Python
python_students = {"Rahul", "Aman", "Priya", "Karan", "Neha"}

# Create a set of students learning Java
java_students = {"Priya", "Karan", "Rohit", "Simran"}

# Print both sets
print("Python Students:", python_students)
print("Java Students:", java_students)
```

```python
# union() combines both sets
# It includes every student, but removes duplicates
either_language = java_students.union(python_students)

print("Students learning Python or Java:", either_language)
```

```python
# intersection() finds students who are present in BOTH sets
both_languages = python_students.intersection(java_students)

print("Students learning both:", both_languages)
```

```python
# difference() finds items that are in the first set
# but NOT in the second set
only_python = python_students.difference(java_students)

print("Students learning only Python:", only_python)
```

```python
# Find students who are in the Java set
# but NOT in the Python set
only_java = java_students.difference(python_students)

print("Students learning only Java:", only_java)
```

```python
# symmetric_difference() finds students who belong
# to ONLY ONE of the two sets
# Students in both groups are excluded
exactly_one_group = python_students.symmetric_difference(java_students)

print("Students in exactly one group:", exactly_one_group)
```

```python
# add() adds ONE new item to a set
python_students.add("Vikram")

print("Python Students after adding Vikram:", python_students)
```

```python
# remove() removes a specific student
# It gives an error if the student does not exist
python_students.remove("Vikram")

print("Python Students after removing Vikram:", python_students)
```

```python
# discard() removes a specific student
python_students.discard("Aman")

print("Python Students after discard():", python_students)

# The important difference:
# discard() does NOT give an error if the item doesn't exist
python_students.discard("John")

print("No error even though John was not in the set")
```
