```python
student = "python programming for data science"
```

```python
# Print the complete string
print(student)
```

```python
# Indexing: Access a character using its position
print(student[0])
```

```python
# Negative indexing: Access a character from the end
print(student[-1])
```

```python
# Slicing: Extract characters from the beginning up to a specified position
print(student[:6])
```

```python
# Slicing: Extract characters from a specified position to the end
print(student[-7:])
```

```python
# Slicing with a step of -1: Reverse the string
print(student[::-1])
```

```python
# upper(): Convert all letters in the string to uppercase
print(student.upper())
```

```python
# lower(): Convert all letters in the string to lowercase
print(student.lower())
```

```python
# title(): Convert the first character of each word to uppercase
print(student.title())
```

```python
# count(): Count how many times a specified character or substring occurs
print(student.count("a"))
```

```python
# find(): Find the starting position of the first occurrence of specified text
print(student.find("programming"))
```

```python
# replace(): Replace specified text with new text
print(student.replace("data science", "artificial intelligence"))
```

```python
# split(): Split the string into a list of smaller strings
print(student.split())
```
