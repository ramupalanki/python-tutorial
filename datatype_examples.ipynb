```python
#assigning integer values, printing and then displaying its type
age=25
print(age)
print(type(age))
```

```python
#assigning floating value, printing and then displaying its type
salary=100000.50
print(salary)
print(type(salary))
```

```python
#assigning string value, printing and then displaying its type
name="Ramu"
print(name)
print(type(name))
```

```python
#assigning boolean values, printing and then displaying its type
like_python=True
print(like_python)
print(type(like_python))
```

```python
#creating list data type, printing and then displaying its type
fruits=["banana","Mango","Apple"]
print(fruits)
print(type(fruits))
```

```python
#creating tuple data type, printing and then displaying its type
names=("Ramu","Ravi","Raju")
print(names)
print(type(names))
```

```python
#creating set data type, printing and then displaying its type
types={"test", "oneday","T20"}
print(types)
print(type(types))
```

```python
#creating dictionary data type, printing and then displaying its type
student_info ={"name":"Ramu","company":"XYZ"}
print(student_info)
print(type(student_info))
```

```python
#converting string to integer
print(int("100"))
```

```python
#converting string to float
print(float("45.67"))
```

```python
#converting integer to string
print(str(500))
```

```python
#converting integer to boolean
print(bool(1))
```

```python
#converting tuple to list
print(list((1, 2, 3)))
```

```python
#converting list to tuple
print(tuple([1, 2, 3]))
```

```python
##converting list to set
print(set([1, 2, 2, 3]))
```
