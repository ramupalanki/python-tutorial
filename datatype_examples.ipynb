```python
#converting string to integer
# 1. int("100")
# Before: "100" is a string (str).
# Conversion: int() converts a numeric string into an integer.
# After: 100 is an integer (int).
# Result: The value can now be used in integer arithmetic.
print(int("100"))
```

```python
#converting string to float
# 2. float("45.67")
# Before: "45.67" is a string (str).
# Conversion: float() converts a numeric string into a floating-point number.
# After: 45.67 is a float (float).
# Result: The value can now be used in decimal-number calculations.

print(float("45.67"))
```

```python
#converting integer to string
# 3. str(500)
# Before: 500 is an integer (int).
# Conversion: str() converts the number into text.
# After: "500" is a string (str).
# Result: The value is now treated as text rather than a number.
print(str(500))
```

```python
#converting integer to boolean
# 4. bool(1)
# Before: 1 is an integer (int).
# Conversion: bool() converts the value into a Boolean value.
# After: 1 becomes True because non-zero numbers are considered True.
# Result: The converted value can be used in Boolean conditions.
print(bool(1))
```

```python
#converting tuple to list
# 5. list((1, 2, 3))
# Before: (1, 2, 3) is a tuple (tuple).
# Conversion: list() converts the tuple into a list.
# After: [1, 2, 3] is a list (list).
# Result: The values are now stored in a mutable collection.
print(list((1, 2, 3)))
```

```python
#converting list to tuple
# 6. tuple([1, 2, 3])
# Before: [1, 2, 3] is a list (list).
# Conversion: tuple() converts the list into a tuple.
# After: (1, 2, 3) is a tuple (tuple).
# Result: The values are now stored in an immutable sequence.
print(tuple([1, 2, 3]))
```

```python
# 7. set([1, 2, 2, 3])
# Before: [1, 2, 2, 3] is a list (list) containing a duplicate 2.
# Conversion: set() converts the list into a set.
# After: {1, 2, 3} is a set (set).
# Result: Duplicate values are automatically removed.
print(set([1, 2, 2, 3]))
```
